## Perform statistical test

### Subtask:
Conduct a Chi-Squared test to determine if the observed distribution of the last two digits from the 3-year data deviates significantly from a uniform distribution.


**Reasoning**:
Import the necessary function for the Chi-Squared test, calculate expected frequencies, and perform the test.



```python
from scipy.stats import chisquare
import numpy as np

# Define observed frequencies from the frequency_counts_threeyear Series and sort them by index
observed_frequencies_threeyear = frequency_counts_threeyear.sort_index().values

# Calculate the total number of observations
total_observations_threeyear = observed_frequencies_threeyear.sum()

# Create an array to hold observed frequencies for all 100 possible digits (0-99), initializing with zeros.
observed_frequencies_all_threeyear = np.zeros(100)
for digit, count in frequency_counts_threeyear.items():
    observed_frequencies_all_threeyear[digit] = count

# Calculate the expected frequencies assuming a uniform distribution (00-99)
expected_frequencies_all_threeyear = np.full(100, total_observations_threeyear / 100)

# Perform the Chi-Squared test
chi_squared_statistic_threeyear, p_value_threeyear = chisquare(f_obs=observed_frequencies_all_threeyear, f_exp=expected_frequencies_all_threeyear)

# Print the results
print("Chi-Squared Statistic (3 years):", chi_squared_statistic_threeyear)
print("P-value (3 years):", p_value_threeyear)
```

## Interpret results

### Subtask:
Explain the results of the statistical test in the context of the distribution of the last two digits from the 3-year data.


**Reasoning**:
Analyze the p-value from the Chi-Squared test for the 3-year data and interpret the results in Thai.



```python
# Analyze the p-value for the 3-year data
alpha = 0.05  # Common significance level

print(f"สถิติ Chi-Squared (3 ปี): {chi_squared_statistic_threeyear:.2f}")
print(f"ค่า P-value (3 ปี): {p_value_threeyear:.10f}") # แสดงทศนิยมมากขึ้นเพื่อความชัดเจน
print(f"ระดับนัยสำคัญ (alpha): {alpha}")

print("\nผลการทดสอบทางสถิติ:")
if p_value_threeyear < alpha:
    print("ค่า P-value น้อยกว่าระดับนัยสำคัญ (0.05)")
    print("เราปฏิเสธสมมติฐานว่าง (Null Hypothesis)")
    print("มีหลักฐานทางสถิติที่ชัดเจนบ่งชี้ว่าการแจกแจงของเลขท้ายสองตัวไม่เป็นแบบสม่ำเสมอ (Uniform Distribution)")
else:
    print("ค่า P-value มากกว่าระดับนัยสำคัญ (0.05)")
    print("เราไม่สามารถปฏิเสธสมมติฐานว่างได้")
    print("ไม่มีหลักฐานทางสถิติเพียงพอที่จะสรุปได้ว่าการแจกแจงของเลขท้ายสองตัวแตกต่างจากการแจกแจงแบบสม่ำเสมอ")

print("\nการตีความผลลัพธ์:")
if p_value_threeyear < alpha:
    print("จากการทดสอบ Chi-Squared พบว่าการแจกแจงของเลขท้ายสองตัวของราคาปิดดัชนี SET ในช่วง 3 ปีที่ผ่านมานั้น **ไม่เป็นไปตามการแจกแจงแบบสม่ำเสมอ** อย่างมีนัยสำคัญทางสถิติ")
    print("นี่แสดงว่าเลขท้ายสองตัวบางตัวมีแนวโน้มที่จะปรากฏบ่อยกว่าหรือน้อยกว่าที่คาดหวังจากการสุ่มแบบสม่ำเสมอ")
else:
    print("จากการทดสอบ Chi-Squared ที่ระดับนัยสำคัญ 0.05 เราไม่มีหลักฐานทางสถิติเพียงพอที่จะสรุปได้ว่าการแจกแจงของเลขท้ายสองตัวของราคาปิดดัชนี SET ในช่วง 3 ปีแตกต่างจากการแจกแจงแบบสม่ำเสมอ")
    print("อย่างไรก็ตาม ควรพิจารณาผลการทดสอบนี้ร่วมกับข้อมูลอื่นๆ และการวิเคราะห์เพิ่มเติม")
```
