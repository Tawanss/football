```python
# เลือก column ที่จะใช้
selected_features = ['Season', 'MatchWeek',
    'Date', 'HomeTeam', 'AwayTeam', 'FullTimeHomeTeamGoals', 'FullTimeAwayTeamGoals','FullTimeResult','HomeTeamShots', 'AwayTeamShots',
    'HomeTeamShotsOnTarget', 'AwayTeamShotsOnTarget',
    'HomeTeamCorners', 'AwayTeamCorners',
    'B365HomeTeam', 'B365Draw', 'B365AwayTeam','B365Over2.5Goals', 'B365Under2.5Goals',


]
df_selec = df[selected_features].copy() # ใช้ .copy()
```

```python
#สร้างคอลลัมใหม่ รวมประตูที่ทำได้ใน 1 เกม
df_selec["goalsaggregate"] = df_selec["FullTimeHomeTeamGoals"] + df_selec["FullTimeAwayTeamGoals"]
```

```python
df_new = df_selec[df_selec['Season'] >= '2002-2003'].copy()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate average goals per season
average_goals_per_season = df_new.groupby('Season')['goalsaggregate'].mean()

# Plot the average goals per season
plt.figure(figsize=(12, 7))
average_goals_per_season.plot(kind='line', marker='o')
plt.title('Average Total Goals per Premier League Season')
plt.xlabel('Season')
plt.ylabel('Average Total Goals')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python


# Calculate average shots on target per season for home and away teams
average_shots_on_target = df_new.groupby('Season')[['HomeTeamShotsOnTarget', 'AwayTeamShotsOnTarget']].mean()

# Plot the average shots on target per season
plt.figure(figsize=(12, 7))
average_shots_on_target.plot(kind='line', marker='o')
plt.title('Average Shots On Target per Premier League Season (Home vs Away)')
plt.xlabel('Season')
plt.ylabel('Average Shots On Target')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Calculate average shots on target per season for home and away teams
average_shots = df_new.groupby('Season')[['HomeTeamShots', 'AwayTeamShots']].mean()

# Plot the average shots on target per season
plt.figure(figsize=(12, 7))
average_shots.plot(kind='line', marker='o')
plt.title('Shots (Home vs Away)')
plt.xlabel('Season')
plt.ylabel('Average Shots')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Count the occurrences of each full time result
result_counts = df_new['FullTimeResult'].value_counts().sort_index()

# Define labels for the results
result_labels = {'H': 'Home Win', 'A': 'Away Win', 'D': 'Draw'}
result_counts.index = result_counts.index.map(result_labels)

# Plot the distribution of results as a pie chart
plt.figure(figsize=(8, 8))
plt.pie(result_counts.values, labels=result_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
plt.title('Distribution of Full Time Match Results (2002-2003 onwards)')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()

# Print the counts
print("Distribution of Match Results:")
print(result_counts)
```

# Task
Analyze the average total goals per season for the "Big 6" teams in the provided dataframe `df_new` starting from the year 2002 and visualize the trend using a line plot.

## Identify big 6 teams

### Subtask:
Define a list of the Big 6 team names.


**Reasoning**:
Define a list containing the names of the Big 6 teams as they appear in the dataframe.



```python
big_6_teams = ['Arsenal', 'Chelsea', 'Liverpool', 'Man City', 'Man United', 'Tottenham']
```

## Filter data

### Subtask:
Filter `df_new` to include only matches where either the home or away team is in the Big 6 list.


**Reasoning**:
Filter the dataframe to include only matches involving the Big 6 teams.



```python
big_6_teams = ['Arsenal', 'Chelsea', 'Liverpool', 'Man City', 'Man United', 'Tottenham']
df_big_6 = df_new[df_new['HomeTeam'].isin(big_6_teams) | df_new['AwayTeam'].isin(big_6_teams)].copy()
```

## Calculate average goals

### Subtask:
Group the filtered data by season and team (considering both home and away matches for each team in a season) and calculate the average `goalsaggregate`.


**Reasoning**:
Filter for home and away matches of Big 6 teams, rename columns, concatenate, and then group by season and team to calculate average goals aggregate.



```python
df_home_goals = df_big_6[df_big_6['HomeTeam'].isin(big_6_teams)][['Season', 'HomeTeam', 'goalsaggregate']].rename(columns={'HomeTeam': 'Team'})
df_away_goals = df_big_6[df_big_6['AwayTeam'].isin(big_6_teams)][['Season', 'AwayTeam', 'goalsaggregate']].rename(columns={'AwayTeam': 'Team'})
df_combined_goals = pd.concat([df_home_goals, df_away_goals])
average_goals_per_season_per_team = df_combined_goals.groupby(['Season', 'Team'])['goalsaggregate'].mean().reset_index()
```

## Plot average goals per season for each big 6 team

### Subtask:
Create a line plot showing the average total goals per season for each of the Big 6 teams.


**Reasoning**:
Create a line plot showing the average total goals per season for each of the Big 6 teams and display it.



```python
plt.figure(figsize=(14, 8))
sns.lineplot(data=average_goals_per_season_per_team, x='Season', y='goalsaggregate', hue='Team', marker='o')
plt.title('Average Total Goals per Season for Big 6 Teams')
plt.xlabel('Season')
plt.ylabel('Average Total Goals')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```
